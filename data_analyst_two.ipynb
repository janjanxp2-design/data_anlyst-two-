```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import pandas as pd
import seaborn as sns

np.random.seed(42)
n_rows = 100000
data = {
    'Region': np.random.choice(['North', 'South', 'East', 'West', 'Central'], n_rows),
    'Category': np.random.choice(['Electronics', 'Clothing', 'Furniture'], n_rows),
    'Quarter': np.random.choice(['Q1', 'Q2', 'Q3', 'Q4'], n_rows),
    'Total_Revenue': np.random.randint(5000, 50000, n_rows),
    'Shipping_Cost': np.random.randint(100, 1500, n_rows),
    'Ad_Spend': np.random.randint(500, 5000, n_rows),
    'Net_Profit': np.random.randint(1000, 15000, n_rows),
    'Units_Sold': np.random.randint(100, 2000, n_rows),
    'Operational_Cost': np.random.randint(2000, 20000, n_rows),
    'Customer_Count': np.random.randint(50, 500, n_rows)
}
df = pd.DataFrame(data)
df
```

```python
region_count=df["Region"].value_counts()
plt.figure(figsize=(5,4))
count=plt.bar(region_count.index,region_count.values)
plt.grid(axis='y', linestyle='--', alpha=0.7)
plt.ylim(0,max(region_count)*1.2)
plt.bar_label(count,padding=2)
plt.title("Bar Plot OF Regions")
plt.xlabel("Regions")
plt.ylabel("Count")
plt.show()
```

```python
category_count=df["Category"].value_counts()
plt.pie(category_count,labels=category_count.index,autopct="%1.1f%%",startangle=90,wedgeprops={'edgecolor': 'black', 'linewidth': 1},
       textprops={'color': 'black', 'fontsize': 11})
plt.title("CATEGORIES USED IN COMANY")
plt.show()
```

```python
import matplotlib.pyplot as plt
region_revenue = df.groupby("Region")["Total_Revenue"].sum()
plt.figure(figsize=(8,5))
plt.ticklabel_format(style='plain', axis='y')
bars = plt.bar(region_revenue.index, region_revenue.values, color='teal', edgecolor='black')
plt.bar_label(bars, labels=[f'{x:,.0f}' for x in region_revenue.values], padding=3)
plt.grid(axis='y', linestyle='--', alpha=0.5)
plt.title("Region Wise Total Revenue", fontsize=15, fontweight='bold')
plt.xlabel("Region", fontsize=10)
plt.ylabel("Total Revenue ($)", fontsize=10)
plt.ylim(0, max(region_revenue.values) * 1.15)
plt.show()
```

```python
qute=df['Quarter']
import matplotlib.pyplot as plt
plt.figure(figsize=(5,3))
count=plt.hist(qute,bins=5,color='skyblue', edgecolor='black')
plt.title('histogram of Quarters')
plt.ylabel('count')
plt.xlabel('Quarters')
plt.show()
```

```python
plt.boxplot(df['Total_Revenue'])
plt.title('boxplot of Totel Revenue')
plt.ylabel("Value")
plt.xlabel("Total Revenue")
plt.show()
```

```python
category_revenue=df.groupby('Category')['Total_Revenue'].sum()
plt.figure(figsize=(5,3))
plt.ticklabel_format(style='plain', axis='y')
bars=plt.bar(category_revenue.index,category_revenue.values,color="green",edgecolor='black')
plt.bar_label(bars, labels=[f'{x:,.0f}' for x in category_revenue.values], padding=3)
plt.ylim(0,max(category_revenue)*1.2)
plt.title('Total Reveue Category wise')
plt.xlabel('CATEGORY')
plt.ylabel('TOTAL REVEUE')
plt.show()
```

```python
plt.figure(figsize=(8,5))
sns.histplot(data=df,x='Shipping_Cost',bins=250,color='coral')
plt.title("DESCRIBING SHIPPING COST (viwe histplot in sns)")
plt.show()
```

```python
df_corr=df[['Total_Revenue','Shipping_Cost','Ad_Spend','Net_Profit','Units_Sold']].corr()
plt.figure(figsize=(8,5))
sns.heatmap(df_corr,annot=True,cmap="YlGnBu", fmt=".2f")
plt.title('HEATMAP OF DATA')
plt.show()
```
